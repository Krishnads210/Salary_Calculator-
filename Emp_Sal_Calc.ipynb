```python
employees = {
    "emp001": {"name": "Krishna", "salary_per_day": 1500, "days_present": 22, "late_days": 3},
    "emp002": {"name": "Anil",    "salary_per_day": 1200, "days_present": 22, "late_days": 0},
    "emp003": {"name": "Sravani", "salary_per_day": 1800, "days_present": 22, "late_days": 5},
    "emp004": {"name": "Meera",   "salary_per_day": 2000, "days_present": 22, "late_days": 1},
    "emp005": {"name": "Rahul",   "salary_per_day": 1350, "days_present": 22, "late_days": 4},
}
```

```python
# Add a new Employee
emp_id = input("Please enter employee id:")
name = input("Please enter employee name:")
salary_per_day = input("Please enter employee salary per day:")
days_present = input("Please enter the no. of days the employee was present:")
late_days = input("Please enter the no. of days the employee was late:")
```

```python
if emp_id not in employees:
    employees[emp_id] = {"name":name,"salary_per_day":int(salary_per_day), "days_present": int(days_present),"late_days":int(late_days)}
else:
    print(emp_id + "ID already exists!")
```

```python
Salaries = []
for i in employees.values():
    # Gross Salary Calculation
    i["Gross_salary"] = i["salary_per_day"] * i["days_present"]
    # Calculating deductions
    if i['late_days'] >= 0 and i['late_days'] <= 2:
        i["Deduction"] = 0
    elif i['late_days'] >= 3 and i['late_days'] <= 5:
        i["Deduction"] = 0.05*i['Gross_salary']
    elif i['late_days'] >= 6 and i['late_days'] <= 10:
        i["Deduction"] = 0.1*i['Gross_salary']
    elif i['late_days'] >10 :
        i["Deduction"] = 0.2*i['Gross_salary']
    # Final Deduction
    i["Final_salary"] = i["Gross_salary"] - i["Deduction"]
    # Highest Salary Calculation
    Salaries.append(i["Final_salary"])
    Highest_salary = sorted(Salaries)[-1]
    #Lowest Salary Calculation
    Lowest_salary = sorted(Salaries)[0]
    #Total Company payroll
    Total_company_payroll = sum(Salaries)


# Employees details
print(employees)
# Company summary
print("The highest salary in the company is " + str(Highest_salary))
print("The lowest salary in the company is " + str(Lowest_salary))
print("Total payroll in the company is " + str(Total_company_payroll))
```

```python
Highest_salary_employee = []
Lowest_salary_employee = []
for i in employees.values():
    if i['Final_salary'] == Highest_salary:
        Highest_salary_employee.extend([("Highest salary employee",i['name'],i['Final_salary'])])
for i in employees.values():
    if i['Final_salary'] == Lowest_salary:
        Lowest_salary_employee.extend([("Lowest salary employee",i['name'],i['Final_salary'])])
print(Highest_salary_employee)
print(Lowest_salary_employee)

        
```
